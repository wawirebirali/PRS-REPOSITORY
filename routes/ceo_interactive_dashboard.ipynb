```python
budget_kpi_query = """

SELECT

COALESCE(SUM(allocated_amount),0) AS company_budget

FROM company_budget

"""


budget_kpi_df = pd.read_sql(
    budget_kpi_query,
    conn
)


budget_kpi_df
```

```python
budget_summary_query = """

SELECT

    COALESCE(
        (SELECT SUM(allocated_amount)
         FROM company_budget),
        0
    ) AS company_budget,


    COALESCE(
        (SELECT SUM(total_amount)
         FROM purchase_orders),
        0
    ) AS budget_used,


    (
        COALESCE(
            (SELECT SUM(allocated_amount)
             FROM company_budget),
            0
        )

        -

        COALESCE(
            (SELECT SUM(total_amount)
             FROM purchase_orders),
            0
        )
    ) AS balance

"""


budget_summary_df = pd.read_sql(
    budget_summary_query,
    conn
)


budget_summary_df
```

```python
import plotly.graph_objects as go
import json


# Get budget data
budget = budget_summary_df.iloc[0]


# Create figure
fig = go.Figure()


fig.add_trace(
    go.Bar(
        x=["Company Budget"],
        y=[budget.company_budget],
        name="Allocated Budget"
    )
)


fig.add_trace(
    go.Bar(
        x=["Company Budget"],
        y=[budget.budget_used],
        name="Budget Used"
    )
)


fig.update_layout(

    title="Budget vs Spending Analysis",

    xaxis_title="",

    yaxis_title="Amount (UGX)",

    barmode="group",

    template="plotly_white"

)


# Display chart in notebook
fig.show()


# Convert Plotly chart to JSON
budget_json = fig.to_json()


# Display JSON
print(budget_json)


# Save JSON file
with open("budget_spending_analysis.json", "w") as file:
    file.write(budget_json)


print("✅ budget_spending_analysis.json created successfully")
```
